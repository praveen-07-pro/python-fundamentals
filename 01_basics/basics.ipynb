## 2. Variables

```python
name = "Praveen"
age = 20
height = 5.9
is_student = True

print(name, age, height, is_student)

# multiple assignment
a, b, c = 1, 2, 3
x = y = z = 0
print(a, b, c, x, y, z)

# swap
a, b = b, a
print(a, b)
```

## 3. Data types

```python
print(type(10))        # int
print(type(3.14))      # float
print(type("hi"))      # str
print(type(True))      # bool
print(type(None))      # NoneType
print(type(2 + 3j))    # complex
print(type([1, 2]))    # list
print(type((1, 2)))    # tuple
print(type({1, 2}))    # set
print(type({"a": 1}))  # dict
```

## 4. Type casting

```python
print(int("25") + 5)
print(float(3))
print(str(100) + " rupees")
print(bool(0), bool(1), bool(""), bool("a"))
print(list("abc"))
print(int(7.9))   # truncates, does not round
```

## 6. Strings

```python
s = "Python Programming"

print(len(s))
print(s[0], s[-1])        # indexing
print(s[0:6])             # slicing
print(s[::2])             # step
print(s[::-1])            # reverse

print(s.upper(), s.lower(), s.title(), s.swapcase())
print(s.replace("Python", "Java"))
print(s.split())
print("-".join(["a", "b", "c"]))
print("  trim me  ".strip())
print(s.find("gram"), s.count("o"))
print(s.startswith("Py"), s.endswith("ing"))
print("abc".isalpha(), "123".isdigit(), "abc123".isalnum())
print("hi " * 3)
```

## 7. String formatting

```python
name, marks = "Praveen", 92.456

print(f"{name} scored {marks:.2f}")        # f-string (best)
print("{} scored {}".format(name, marks))   # format()
print("%s scored %.1f" % (name, marks))     # old style
print(f"{10 + 5=}")                          # debug style
print(f"{1000000:,}")                        # thousands separator
print("Tab\tseparated and new\nline")       # escape characters
print(r"raw string \n stays as it is")
```
