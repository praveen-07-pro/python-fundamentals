# 02 · Operators

## 1. Arithmetic

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b)    # true division
print(a // b)   # floor division
print(a % b)    # modulus
print(a ** b)   # power
print(divmod(a, b))
```

## 2. Comparison

```python
a, b = 10, 20
print(a == b, a != b, a > b, a < b, a >= b, a <= b)
print(1 < 5 < 10)   # chained comparison
```

## 3. Logical

```python
x, y = True, False
print(x and y, x or y, not x)

age = 22
print(age > 18 and age < 60)

# short-circuit: returns the deciding operand
print(0 or "default")
print("a" and "b")
```

## 4. Assignment

```python
n = 10
n += 5;  print(n)
n -= 3;  print(n)
n *= 2;  print(n)
n /= 4;  print(n)
n //= 2; print(n)
n %= 2;  print(n)
n = 3
n **= 3; print(n)
```

## 5. Bitwise

```python
a, b = 12, 10          # 1100, 1010
print(a & b)           # AND  -> 8
print(a | b)           # OR   -> 14
print(a ^ b)           # XOR  -> 6
print(~a)              # NOT  -> -13
print(a << 2)          # left shift  -> 48
print(a >> 2)          # right shift -> 3
print(bin(a), bin(b))

# tricks
n = 8
print("power of 2:", n & (n - 1) == 0)
print("odd" if 7 & 1 else "even")
```

## 6. Membership and identity

```python
print("a" in "apple", "z" not in "apple")
print(3 in [1, 2, 3])

p = [1, 2]; q = [1, 2]; r = p
print(p == q)   # same value
print(p is q)   # different objects
print(p is r)   # same object
```

## 7. Operator precedence

```python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(2 ** 3 ** 2)     # 512 (right to left)
print(not True or True)
```
