# 05 · Loops

## 1. for loop and range()

```python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 11, 2):     # start, stop, step
    print(i, end=" ")
print()

for i in range(5, 0, -1):     # countdown
    print(i, end=" ")
print()

for ch in "Python":
    print(ch, end=" ")
```

## 2. Looping over collections

```python
fruits = ["apple", "banana", "mango"]

for f in fruits:
    print(f)

for i, f in enumerate(fruits, start=1):
    print(i, f)

names = ["A", "B", "C"]
marks = [90, 80, 70]
for n, m in zip(names, marks):
    print(n, m)

d = {"x": 1, "y": 2}
for k, v in d.items():
    print(k, v)
```

## 3. while loop

```python
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()

# sum of digits
n, total = 1234, 0
while n > 0:
    total += n % 10
    n //= 10
print(total)
```

## 4. break, continue, pass, else

```python
for i in range(1, 10):
    if i == 3:
        continue      # skip 3
    if i == 7:
        break         # stop at 7
    print(i, end=" ")
print()

# else runs only if loop was NOT broken
for i in range(2, 10):
    if i == 20:
        break
else:
    print("loop finished without break")

for i in range(3):
    pass   # placeholder
```

## 5. Nested loops and patterns

```python
for i in range(1, 6):
    print("*" * i)

print()
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end="\t")
    print()
```

## 6. Common programs

```python
# Factorial
n, fact = 5, 1
for i in range(1, n + 1):
    fact *= i
print(fact)

# Fibonacci
a, b = 0, 1
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
print()

# Prime check
num = 29
is_prime = num > 1 and all(num % i for i in range(2, int(num ** 0.5) + 1))
print(num, "prime" if is_prime else "not prime")

# Palindrome
w = "madam"
print(w == w[::-1])
```
