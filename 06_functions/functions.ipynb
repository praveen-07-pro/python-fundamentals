# 06 · Functions

## 1. Defining and calling

```python
def greet(name):
    """Return a greeting."""
    return f"Hello, {name}!"

print(greet("Praveen"))
print(greet.__doc__)
```

## 2. Parameters

```python
def power(base, exp=2):          # default argument
    return base ** exp

print(power(5))
print(power(2, 10))
print(power(exp=3, base=2))      # keyword arguments

def add_all(*args):              # any number of positional args
    return sum(args)
print(add_all(1, 2, 3, 4))

def show(**kwargs):              # any number of keyword args
    for k, v in kwargs.items():
        print(k, "=", v)
show(name="Praveen", age=20)

def mixed(a, b=0, *args, **kwargs):
    print(a, b, args, kwargs)
mixed(1, 2, 3, 4, x=5)
```

## 3. Return values

```python
def min_max(lst):
    return min(lst), max(lst)    # returns a tuple

lo, hi = min_max([4, 9, 1, 7])
print(lo, hi)

def no_return():
    pass
print(no_return())               # None
```

## 4. Scope

```python
x = "global"

def f():
    x = "local"
    print(x)

f()
print(x)

counter = 0
def inc():
    global counter
    counter += 1
inc(); inc()
print(counter)

def outer():
    n = 0
    def inner():
        nonlocal n
        n += 1
        return n
    return inner()
print(outer())
```

## 5. Lambda, map, filter, sorted

```python
sq = lambda x: x ** 2
print(sq(6))

nums = [1, 2, 3, 4, 5, 6]
print(list(map(lambda x: x * 2, nums)))
print(list(filter(lambda x: x % 2 == 0, nums)))

people = [("A", 30), ("B", 22), ("C", 27)]
print(sorted(people, key=lambda p: p[1]))
```

## 6. Recursion

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print(factorial(5))
print([fib(i) for i in range(10)])
```

## 7. Type hints and useful built-ins

```python
def area(r: float) -> float:
    return 3.14159 * r * r

print(area(3))
print(abs(-5), round(3.14159, 2), pow(2, 3), len("abc"))
print(any([0, 1]), all([1, 1]), sorted([3, 1, 2], reverse=True))
```

## 8. Mini programs

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def reverse_string(s):
    return s[::-1]

def count_vowels(s):
    return sum(1 for c in s.lower() if c in "aeiou")

print([n for n in range(30) if is_prime(n)])
print(reverse_string("python"), count_vowels("Data Science"))
```
