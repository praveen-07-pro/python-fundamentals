## 1. List (ordered, mutable)

```python
nums = [5, 2, 8, 1]

nums.append(9)
nums.insert(1, 100)
nums.extend([7, 3])
print(nums)

nums.remove(100)
print(nums.pop(), nums.pop(0))
print(nums)

nums.sort();               print(nums)
nums.sort(reverse=True);   print(nums)
nums.reverse();            print(nums)
print(sorted(nums), len(nums), sum(nums), min(nums), max(nums))
print(nums.index(8), nums.count(2))

print(nums[1:4], nums[::-1])
copy = nums.copy()

# 2D list
matrix = [[1, 2, 3], [4, 5, 6]]
print(matrix[1][2])
```

## 2. Tuple (ordered, immutable)

```python
t = (10, 20, 30, 20)
print(t[0], t[-1], t[1:3])
print(t.count(20), t.index(30))

single = (5,)        # comma needed for single element
a, b, *rest = (1, 2, 3, 4)   # unpacking
print(a, b, rest)
# t[0] = 99   # TypeError: tuples cannot be changed
```

## 3. Set (unordered, unique)

```python
s = {1, 2, 3, 3, 2}
print(s)

s.add(4)
s.discard(10)    # no error if missing
s.remove(1)
print(s)

a, b = {1, 2, 3, 4}, {3, 4, 5}
print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
print(a ^ b)   # symmetric difference

print(list(set([1, 1, 2, 3, 3])))   # remove duplicates
```

## 4. Dictionary (key-value)

```python
student = {"name": "Praveen", "age": 20, "dept": "AI&DS"}

print(student["name"])
print(student.get("cgpa", "N/A"))   # safe access

student["cgpa"] = 8.5
student["age"] = 21
del student["dept"]
print(student)

print(student.keys())
print(student.values())
print(student.items())

for k, v in student.items():
    print(k, "->", v)

student.update({"city": "Trichy"})
print(student.pop("city"))
print("name" in student)

# word count
text = "to be or not to be"
count = {}
for w in text.split():
    count[w] = count.get(w, 0) + 1
print(count)
```

## 5. Comprehensions

```python
squares = [x ** 2 for x in range(1, 6)]
evens = [x for x in range(20) if x % 2 == 0]
sq_dict = {x: x ** 2 for x in range(1, 6)}
unique_len = {len(w) for w in ["a", "bb", "cc", "ddd"]}
print(squares, evens, sq_dict, unique_len)
```
