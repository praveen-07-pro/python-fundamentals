# 03 · Control Flow

## 1. if / elif / else

```python
marks = 78

if marks >= 90:
    print("A+")
elif marks >= 75:
    print("A")
elif marks >= 50:
    print("B")
else:
    print("Fail")
```

## 2. Nested if

```python
age, has_id = 20, True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Show ID")
else:
    print("Underage")
```

## 3. Ternary (one-line if)

```python
n = 7
print("even" if n % 2 == 0 else "odd")
```

## 4. Truthy and falsy

```python
for v in [0, 1, "", "a", [], [0], None, 0.0]:
    print(repr(v), "->", bool(v))

items = []
if not items:
    print("list is empty")
```

## 5. match-case (Python 3.10+)

```python
day = 3

match day:
    case 1:
        print("Mon")
    case 2:
        print("Tue")
    case 3:
        print("Wed")
    case 6 | 7:
        print("Weekend")
    case _:
        print("Other")
```

## 6. Small programs

```python
# Largest of three
a, b, c = 12, 45, 30
print(max(a, b, c))

# Leap year
year = 2024
print(year, "leap" if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0 else "not leap")

# Simple calculator
x, y, op = 10, 5, "*"
if op == "+": print(x + y)
elif op == "-": print(x - y)
elif op == "*": print(x * y)
elif op == "/": print(x / y)
```
